```python
class Solution:
    def shuffle(self, nums, n):
        result = []
        for i in range(n):
            result.append(nums[i])
            result.append(nums[i + n])
        return result

raw = input().strip()

if raw.startswith('[') and raw.endswith(']'):
    raw = raw[1:-1]

parts = raw.split(',')

nums = [int(x.strip()) for x in parts if x.strip()]

n = int(input().strip())

obj = Solution()
print(obj.shuffle(nums, n))
```
